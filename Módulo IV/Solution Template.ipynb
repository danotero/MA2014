# Reinforcement Learning

Reinforcement learning (RL) is a machine learning method where an AI agent learns to make decisions by trial and error, taking actions within an environment to maximize a cumulative reward over time.

Instead of being explicitly taught the right answers, the agent learns through feedback—much like training a dog with treats or learning to master a video game through practice:

* **Agent:** The learner or decision-maker.
* **Environment:** The world or system the agent interacts with.
* **Action:** The choice the agent makes at any given step.
* **Reward:** The numeric feedback (positive or negative) given after an action to signal whether it was a good move.

Over thousands or millions of iterations, the agent builds a policy—a strategy for choosing the best action in any situation. Reinforcement learning drives key breakthroughs across modern technology, including mastering complex games (like Go and chess), training self-driving cars, controlling robotic systems, and fine-tuning AI language models through human feedback.

## RL Algorithms

<center><img src="RL_taxonomy.jpeg" alt="Drawing" style="width: 700px;"/></center>

The image provides a structured taxonomy of Reinforcement Learning (RL) methods based on Sutton & Barto (2018). The taxonomy splits into two primary paradigms: **Tabular Methods** and **Approximate Methods**.

### Tabular Methods (Finite State/Action Spaces)

* **Model-Free Methods (Learning from interaction)**:
* **Monte Carlo (MC) Methods**: Update value estimates based on complete episode returns without bootstrapping.
* **Temporal-Difference (TD) Methods**: Learn incrementally using partial returns and bootstrapping. Algorithms include **Sarsa** (On-Policy), **Q-Learning** (Off-Policy), and **Expected Sarsa**.
* **Model-Based Methods (Dynamic Programming/Planning)**:
* **Dynamic Programming (DP)**: Requires an explicit model of the environment and relies on **Policy Iteration** and **Value Iteration**.
* **Unified Approaches**: Combines TD interaction with model-based planning (e.g., **Dyna-Q** and integrated planning & learning frameworks).

### Approximate Methods (Function Approximation)

* **Value Function Approximation (VFA)**: Approximates state or action values ($V(s), Q(s,a)$) across continuous or large state spaces. Uses **Linear Methods** or **Neural Networks** (e.g., Deep RL, DQN).
* **Policy Function Approximation (PG)**: Directly parameterizes the policy ($\pi(a\vert{}s)$) using the **Policy Gradient Theorem** and algorithms like **REINFORCE**.
* **Actor-Critic Methods**: Combines value approximation and policy gradient approaches, including architectures like **Advantage Actor-Critic (A2C/A3C)**.

### Core Structural Axes

* **On-Policy vs. Off-Policy**: Distinguishes methods where the policy evaluated/improved is the same as the policy generating the data (e.g., Sarsa) versus methods where they differ (e.g., Q-Learning).
* **Bootstrapping (TD) vs. No Bootstrapping (MC)**: Highlights the trade-off between updating estimates based on other estimated values versus learning exclusively from full, un-bootstrapped trajectories.

# Route Planning

**Route planning in robotics** is the computational process of determining an optimal, collision-free sequence of waypoints or vectors that enables an autonomous robot to navigate from a starting position to a target destination within a given environment.

It forms a critical layer of a robot's navigation architecture, bridging the gap between high-level task execution and low-level motor control.

### Common Algorithmic Approaches

* **Graph and Grid Search:** Algorithms like Dijkstra's, $A^*$, and $D^*$ discretization methods that search structured grid maps or topological graphs for the lowest-cost path.
* **Sampling-Based Planners:** Methods like Rapidly-exploring Random Trees (RRT/RRT*) and Probabilistic Roadmaps (PRM), which build random trees/graphs across continuous spaces, ideal for high-dimensional robotic arms or complex mobile environments.
* **Potential Fields:** Virtual physics models where the destination exerts an attractive force and obstacles exert repulsive forces to guide the robot.
* **Learning-Based Approaches:** Reinforcement learning and neural network policies that learn routing strategies through trial-and-error interaction with the environment.

## The Q-Learning Algorithm

Q-Learning is a model-free, off-policy Temporal Difference (TD) reinforcement learning algorithm designed to find the optimal action-selection policy for any Markov Decision Process (MDP).

### Mathematical Formulation

The objective of Q-Learning is to estimate the action-value function $Q(s, a)$, which represents the expected cumulative discounted reward of taking action $a$ in state $s$ and following an optimal policy thereafter.Bellman Optimality EquationThe theoretical foundation of Q-learning is anchored in the Bellman Optimality Equation for $Q^*(s, a)$:

$$Q^*(s, a) = \sum_{s', r} p(s', r \mid s, a) \left[ r + \gamma \max_{a'} Q^*(s', a') \right]$$

### Temporal Difference (TD) Update Rule

In model-free environments, transition probability distributions $p(s', r \mid s, a)$ are unknown. Q-learning uses sampled experience tuples $(s, a, r, s')$ to update $Q(s, a)$ iteratively:

$$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
